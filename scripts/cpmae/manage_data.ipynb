```python
from huggingface_hub import HfApi

api = HfApi()

repo_id = "haitong-ma/cpmae_ssl_results"

# List all files in the repo
files = api.list_repo_files(repo_id)

# Filter files under r3m/ or vip/
files_to_delete = [
    f for f in files if 'r3m' in f or 'vip' in f
]
```

```python
api.delete_files(
    repo_id=repo_id,
    delete_patterns=files_to_delete,
    commit_message="Remove all r3m and vip folders"
)
```
